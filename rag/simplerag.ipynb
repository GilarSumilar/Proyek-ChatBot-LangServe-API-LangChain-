```python
#PDF reader
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("CV-Gilar_Sumilar.pdf")
docs = loader.load()
```

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
documents = text_splitter.split_documents(docs)
documents[:5]
```

```python
documents
```

```python
import langchain_google_genai
import google.genai

print("langchain_google_genai OK")
print("google.genai OK")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma

# Embedding model
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

# Test embedding dulu
result = embeddings.embed_query("halo dunia")
print(f"Jumlah dimensi embedding: {len(result)}")

# Simpan ke Chroma
db = Chroma.from_documents(
    documents=documents[:15],
    embedding=embeddings
)

print(f"Jumlah dokumen di Chroma: {db._collection.count()}")
```

```python
# Melihat isi data yang tersimpan di Chroma (Vector Database).
data = db.get()

for i, doc in enumerate(data["documents"]):
    print(f"\n=== Dokumen {i+1} ===")
    print(doc)
```

```python
#Mencari dokumen yang paling relevan dengan pertanyaan.
query = "Apa saja pengalaman kerja Gilar Sumilar?"

results = db.similarity_search(query, k=2)

for doc in results:
    print(doc.page_content)
    print("-" * 10)
```

```python
#Menggabungkan hasil pencarian Chroma menjadi satu teks (context) yang nantinya dikirim ke Gemini.
context = "\n\n".join(
    [doc.page_content for doc in results]
)

print(context)
```

```python
# Tahap akhir RAG (Retrieval Augmented Generation).
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

prompt = f"""
Jawab pertanyaan berdasarkan konteks berikut.

Konteks:
{context}

Pertanyaan:
{query}
"""

response = llm.invoke(prompt)

print(response.content)
```
