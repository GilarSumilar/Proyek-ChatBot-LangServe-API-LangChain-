```python
#PDF reader
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("CV-Gilar_Sumilar.pdf")
docs = loader.load()
```

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
documents = text_splitter.split_documents(docs)
documents[:5]
```

```python
documents
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import FAISS

# Embedding model
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

# Test embedding dulu
result = embeddings.embed_query("halo dunia")
print(f"Jumlah dimensi embedding: {len(result)}")

# Simpan embedding ke FAISS
db = FAISS.from_documents(
    documents=documents[:15],
    embedding=embeddings
)

print(f"Jumlah dokumen di FAISS: {db.index.ntotal}")
```

```python
# Membuat Retriever untuk mengambil 2 dokumen paling relevan dari FAISS.
retriever = db.as_retriever(search_kwargs={"k": 2})
```

```python
# Mencari dokumen paling relevan dari FAISS melalui Retriever.
query = "Apa saja pengalaman kerja Gilar Sumilar?"
results = retriever.invoke(query)

for doc in results:
    print(doc.page_content)
    print("-" * 10)
```

```python
# Document Chain: format retrieved documents, build the prompt, and call Gemini.
from operator import itemgetter
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableLambda, RunnablePassthrough
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

document_prompt = ChatPromptTemplate.from_messages([
    ("system", "Jawab pertanyaan hanya berdasarkan konteks berikut. Jika jawaban tidak ada dalam konteks, katakan tidak ditemukan."),
    ("human", "Konteks:\n{context}\n\nPertanyaan: {input}")
])

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

document_chain = (
    RunnableLambda(lambda inputs: {**inputs, "context": format_docs(inputs["context"])})
    | document_prompt
    | llm
)
rag_chain = (
    RunnablePassthrough.assign(context=itemgetter("input") | retriever)
    | document_chain
)
```

```python
response = rag_chain.invoke({"input": query})

print(response.content)
```
